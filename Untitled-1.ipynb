## Lab 7: Text Vectorization — Bag of Words and TF-IDF

### Question 1 — Bag-of-Words Matrix  
Create a 5-document toy corpus related to machine learning:      
D1: "machine learning is useful"      
D2: "machine learning is powerful"          
D3: "deep learning is useful"           
D4: "machine learning uses data"          
D5: "data science is useful"            
•	Using CountVectorizer:           
•	Build the Bag-of-Words matrix.         
•	Extract the vocabulary.         
•	Convert the matrix into a Pandas DataFrame.        
•	Display the document-term matrix with document names as row labels.        
Expected Output:          
Vocabulary and a 5-document Bag-of-Words DataFrame.     


```python
from sklearn.feature_extraction.text import CountVectorizer
import pandas as pd

corpus = [
    "machine learning is useful",
    "machine learning is powerful",
    "deep learning is useful",
    "machine learning uses data",
    "data science is useful"
]

vectorizer = CountVectorizer()
X = vectorizer.fit_transform(corpus)
df_bow = pd.DataFrame(X.toarray(), index=[f"D{i+1}" for i in range(len(corpus))], columns=vectorizer.get_feature_names_out())
print(df_bow)
```

### Question 2 — TF-IDF Matrix and Top-Weighted Term

Using the same 5-document corpus from Question 1:  
Build the TF-IDF matrix using TfidfVectorizer.  
Display the matrix as a Pandas DataFrame.   
Identify the top-weighted term for each document.   
Display the term along with its TF-IDF score.   
Expected Output:   
Document 1 → Term: ______ → TF-IDF: ______  
Document 2 → Term: ______ → TF-IDF: ______   
Document 3 → Term: ______ → TF-IDF: ______   
Document 4 → Term: ______ → TF-IDF: ______     
Document 5 → Term: ______ → TF-IDF: ______   


```python
from sklearn.feature_extraction.text import TfidfVectorizer

tfidf = TfidfVectorizer()
X_tfidf = tfidf.fit_transform(corpus)
df_tfidf = pd.DataFrame(X_tfidf.toarray(), index=[f"D{i+1}" for i in range(len(corpus))], columns=tfidf.get_feature_names_out())
print(df_tfidf.round(3))

for i, row in df_tfidf.iterrows():
    top_term = row.idxmax()
    top_score = row.max()
    print(f"{i} → Term: {top_term} → TF-IDF: {top_score:.3f}")
```

### Question 3 — Pairwise Cosine Similarity

Using the TF-IDF vectors generated in Question 2:  
Calculate pairwise cosine similarity using cosine_similarity().  
Display the similarity matrix as a Pandas DataFrame.  
Identify the most similar pair of documents.  
Identify the least similar pair of documents.  
Expected Output:  
           D1     D2     D3     D4     D5      
D1        1.00   ...    ...    ...    ...   
D2        ...    1.00   ...    ...    ...  
D3        ...    ...    1.00   ...    ...    
D4        ...    ...    ...    1.00   ...   
D5        ...    ...    ...    ...    1.00   

Most Similar: D__ and D__   
Least Similar: D__ and D__  


```python
from sklearn.metrics.pairwise import cosine_similarity
import numpy as np

cos_sim = cosine_similarity(X_tfidf)
df_sim = pd.DataFrame(cos_sim, 
                      index=[f"D{i+1}" for i in range(len(corpus))], 
                      columns=[f"D{i+1}" for i in range(len(corpus))])
df_sim

np.fill_diagonal(cos_sim, 0)
max_pair = divmod(cos_sim.argmax(), cos_sim.shape[1])
cos_sim_no_diag = cos_sim.copy()
np.fill_diagonal(cos_sim_no_diag, np.nan)
min_pair = divmod(np.nanargmin(cos_sim_no_diag), cos_sim_no_diag.shape[1])
print("Most Similar:", f"D{max_pair[0]+1} and D{max_pair[1]+1}")
print("Least Similar:", f"D{min_pair[0]+1} and D{min_pair[1]+1}")

```
